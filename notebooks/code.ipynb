# Mapping

```python
mapping = {
    'GitHub_Profile' : {'Yes' : 1 , 'No' : 0},
    'Leadership_Experience' : {'Yes' : 1 , 'No' : 0},
    'LinkedIn_Profile' : {'Yes' : 1 , 'No' : 0},
    'Placement_Status' : {'Placed' : 1 , 'Not Placed' : 0},
}

for col , mapping in mapping.items():
    df[col] = df[col].map(mapping)
    
```

```python
df_copy = df.copy()
```

```python
df_copy.columns.tolist()
```

# Remove Unnecessary Columns

```python
df_copy = df_copy.iloc[: , :-5]
```

```python
# check
df_copy.columns.to_list()
```

```python
# I removed these columns because these created after interview and  could be a noise to our model
column = ['Resume_Score','Interview_Score' ,'Employability_Score','Attendance_Percentage']
df_copy = df_copy.drop(column , axis=1)
```

```python
df_copy.columns.tolist()
```

```python
# columns that will be encoded
alpha_cols_after_mapping = df_copy.select_dtypes(include='O').columns.tolist()

alpha_cols_after_mapping
```

# Encoding

```python
from sklearn.preprocessing import OneHotEncoder , OrdinalEncoder

encoder = OneHotEncoder(drop='first')

ordinalencoder = OrdinalEncoder(categories=[

    ['Freshman', 'Sophomore', 'Junior', 'Senior'],
    ['Poor', 'Average', 'Good', 'Excellent'],
    ['Basic', 'Intermediate', 'Advanced']

])


df_copy[['University_Year','Academic_Performance','English_Proficiency']] = ordinalencoder.fit_transform(
    df_copy[['University_Year',
        'Academic_Performance',
        'English_Proficiency']]
)
```

```python
encoded = encoder.fit_transform(df_copy[['Major' , 'Gender']])

encoded_df = pd.DataFrame(
    encoded.toarray(),
    columns=encoder.get_feature_names_out(['Major', 'Gender']),
    index=df_copy.index
)

df_copy = pd.concat([df_copy.drop(columns=['Major', 'Gender']), encoded_df], axis=1)
```

```python
df_copy.info()
```

# Train / Test Split

```python
from sklearn.model_selection import train_test_split

x = df_copy.drop(columns= 'Placement_Status')
y = df_copy['Placement_Status']
```

```python
x_train , x_test , y_train , y_test = train_test_split(
    x,
    y,
    test_size=0.1,
    stratify = y,
    random_state=42)
```

```python
y_train.value_counts(normalize=True) * 100
```

# Scailng

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

# Smote

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42 ,sampling_strategy=0.7,k_neighbors=4)

x_train_smote , y_train_smote = smote.fit_resample(x_train_scaled,y_train)
```

```python
y_train_smote.value_counts()
```

# Random Forest

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(random_state=42,n_estimators=500,min_samples_leaf=5,min_samples_split=7,max_features=10)
```

```python
# with smooth
rf.fit(x_train_smote,y_train_smote)
```

```python
y_pred = rf.predict(x_test_scaled)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
# With Balance Data
rf_balanced = RandomForestClassifier(
    n_estimators=500,
    class_weight='balanced',
    random_state=42
)

rf_balanced.fit(x_train_scaled, y_train)
```

```python
y_pred_balanced = rf_balanced.predict(x_test_scaled)
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred_balanced))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_balanced))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_balanced))
```

# Logistic Regression With Smote

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression(random_state=42,max_iter=200)

lr.fit(x_train_smote, y_train_smote)
```

```python
y_pred_lr = lr.predict(x_test_scaled)
```

```python
from sklearn.ensemble import AdaBoostClassifier

ada = AdaBoostClassifier(n_estimators=250 , learning_rate=0.4 , random_state=42)
ada_scaled_smote = AdaBoostClassifier(n_estimators=250 , learning_rate=0.5 , random_state=42)
```

```python
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix

print("Accuracy:", accuracy_score(y_test, y_pred_ada))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_ada))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_ada))
```

```python
# With Smote
print("Accuracy:", accuracy_score(y_test, y_pred_ada_smote))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_ada_smote))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_ada_smote))
```

# Cross-Validation

```python
from sklearn.model_selection import StratifiedKFold

cv = StratifiedKFold(n_splits=5,
                     shuffle=True,
                     random_state=42)
```

```python
from imblearn.pipeline import Pipeline

pipeline_ada = Pipeline([
    ('smote',SMOTE(sampling_strategy=0.7,random_state=42)),
    ('model',AdaBoostClassifier(n_estimators=250,
                                learning_rate=0.5,
                                random_state=42
    ))
])


```

```python
from sklearn.model_selection import cross_validate

cv_results = cross_validate(
    pipeline_ada,
    x_train_scaled,
    y_train,
    cv=cv,
    scoring={
        'accuracy' : 'accuracy',
        'recall' : 'recall',
        'recall_macro': 'recall_macro',
        'f1_macro': 'f1_macro'
    },
    n_jobs=-1
)

print("Mean CV Accuracy:", cv_results['test_accuracy'].mean())
print("Mean CV Recall (Placed):", cv_results['test_recall'].mean())
print("Mean CV Recall Macro:", cv_results['test_recall_macro'].mean())
print("Mean CV Macro F1:", cv_results['test_f1_macro'].mean())
```

```python
print("Accuracy:", cv_results['test_accuracy'])
print("Recall Placed:", cv_results['test_recall'])
print("Recall Macro:", cv_results['test_recall_macro'])
print("Macro F1:", cv_results['test_f1_macro'])
```

# Tuning

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    'model__n_estimators' : [100 , 150 , 200 , 250 , 300 , 400 , 125],
    'model__learning_rate' : [0.05 , 0.1 , 0.25 , 0.5 , 0.75 , 1.0]
}

grid_ada = GridSearchCV(estimator=pipeline_ada,
                         param_grid=param_grid,
                         cv=cv,
                         scoring='recall',
                         n_jobs=-1)
```

```python
grid_ada.fit(x_train_scaled, y_train)

print("Best Parameters:", grid_ada.best_params_)
print("Best CV Accuracy:", grid_ada.best_score_)
```

```python
from sklearn.metrics import confusion_matrix, classification_report

best_ada = grid_ada.best_estimator_

y_pred_best = best_ada.predict(x_test)

cm = confusion_matrix(y_test, y_pred_best)

print("Confusion Matrix:")
print(cm)

print("\nClassification Report:")
print(classification_report(y_test, y_pred_best))
```
